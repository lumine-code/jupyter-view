# Signal Lab
Compare the sample sequence with its mean amplitude.

```python
import numpy as np
samples = np.array([0.2, 0.8, -0.4, 0.5, 1.0])
print(f"Average: {samples.mean():.2f}")
amplitude: float = "high"
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(7, 2.2))
plt.plot(samples, color="#bd881f", marker="o")
plt.title("Sample amplitudes")
plt.grid(alpha=0.2)
plt.show()
```
